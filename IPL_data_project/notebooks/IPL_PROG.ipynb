# 1.which team won ther most matches

````python
df.head()
````

````python
Match_wins=df['match_winner'].value_counts()
````

````python
sns.barplot(y=Match_wins.index,x=Match_wins.values,palette="viridis")
plt.title("Most Match Wins By Teams")
plt.xlabel("Teams")
plt.ylabel("")
````

# 2.Toss Decision trends

````python
sns.countplot(x=df["toss_decision"],palette="rainbow")
plt.title("Toss Decision trends")
plt.savefig("../images/toss_decision.png", dpi=300, bbox_inches="tight")
````

# 3. Toss Winner VS Match Winner

````python
Count= df[df["toss_winner"]== df["match_winner"]]['match_id'].count()
Percentage=(Count*100)/df.shape[0]
Percentage.round(2)
````

# key Player Performance
1 Most "Player of the match" Awards

````python
Player_of_match=df["player_of_the_match"].value_counts().head(10)
sns.barplot(x=Player_of_match.values , y=Player_of_match.index,palette="mako")
plt.title("top 10 players with man of the match awards ")
plt.savefig("../images/top_players.png", dpi=300, bbox_inches="tight")
````

2 Top scorers

````python
high=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
high.plot(kind="barh")
plt.savefig("../images/High_scorer.png", dpi=300, bbox_inches="tight")
````

10 Best bowling Figures

````python
df["highest_wickets"]=df["best_bowling_figure"].apply(lambda x : x.split("--")[0])
df["highest_wickets"]=df["highest_wickets"].astype(int)
k=df.groupby("best_bowling")["highest_wickets"].sum().sort_values(ascending=False).head(10)
sns.barplot(y=k.index,x=k.values,palette="viridis")
````

# Venue Analysis
